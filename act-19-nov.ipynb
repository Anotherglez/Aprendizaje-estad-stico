## 1. Regresión Lineal Múltiple con `statsmodels` (OLS)

Ajustamos un modelo de regresión lineal por Mínimos Cuadrados Ordinarios (OLS) sobre el dataset **Advertising** utilizando las variables predictoras `TV`, `radio` y `newspaper` para estimar las ventas (`sales`):

$$\hat{y} = \hat{\beta}_0 + \hat{\beta}_1 \cdot \text{TV} + \hat{\beta}_2 \cdot \text{radio} + \hat{\beta}_3 \cdot \text{newspaper}$$

```python
import pandas as pd
import statsmodels.api as sm

# Cargar datos
df = pd.read_csv("Advertising.csv")

X = df[["TV", "radio", "newspaper"]]
y = df["sales"]

# Agregar constante para el intercepto (beta 0)
X = sm.add_constant(X)

# Ajuste OLS
modelo = sm.OLS(y, X).fit()
print(modelo.summary())
```

## 2. Remuestreo con Reemplazo (Bootstrapping)

Utilizamos la función `resample` de `sklearn.utils` para generar muestras artificiales del mismo tamaño muestral extraídas con reemplazo ($replace = True$) para analizar la variabilidad de las estimaciones.

```python
from sklearn.utils import resample

# Generar una muestra bootstrap con semilla fija
Xb, yb = resample(X, y, random_state=5)

print("Dimensiones Xb:", Xb.shape)
print("\nPrimeras filas de la muestra bootstrap:")
display(Xb.head())
```

## 3. Ajuste de Regresión Lineal sobre Muestra Bootstrap

Ajustamos un modelo de regresión lineal con `scikit-learn` sobre la muestra bootstrap obtenida.

```python
from sklearn.linear_model import LinearRegression as ls

modelob = ls()
modelob.fit(Xb, yb)

print("Intercepto (beta 0):", modelob.intercept_)
```

## 4. Simulación Bootstrap e Inferencia de Parámetros

Iteramos el procedimiento de remuestreo $B$ veces para acumular los valores estimados del intercepto y calcular su media global.

```python
for i in range (10000) :
     Xb, yb = resample(X, y, random_state=i)

     modelob.fit(Xb, yb)
     

import numpy as np 

print(np.mean(modelob.intercept_))
```
